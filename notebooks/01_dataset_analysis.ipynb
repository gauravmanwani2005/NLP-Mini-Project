# EduMind Notebook 01: Domain-Specific Dataset Analysis & Exploration
This notebook inspects the college academic query dataset, analyzes class balance, sequence length distributions, and token vocabulary.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load datasets
train_df = pd.read_csv('../data/processed/train.csv')
val_df = pd.read_csv('../data/processed/validation.csv')
test_df = pd.read_csv('../data/processed/test.csv')

print(f"Train size: {len(train_df)}")
print(f"Validation size: {len(val_df)}")
print(f"Test size: {len(test_df)}")
print(f"Total samples: {len(train_df) + len(val_df) + len(test_df)}")
```

```python
# Inspect class distribution
print("Train Class Distribution:")
print(train_df['intent'].value_counts())
```

```python
# Plot class balance
plt.figure(figsize=(12, 5))
sns.countplot(data=train_df, x='intent', order=train_df['intent'].value_counts().index, palette='viridis')
plt.xticks(rotation=45, ha='right')
plt.title('Training Set Class Balance across 12 Academic Intents')
plt.tight_layout()
plt.show()
```

```python
# Sentence length distribution
train_df['char_length'] = train_df['text'].apply(len)
train_df['word_count'] = train_df['text'].apply(lambda x: len(x.split()))

print("Summary statistics of query word count:")
print(train_df['word_count'].describe())
```
